```python
# CELL 1: Define Intents and Knowledge Base

## Modified INTENT_DATA to put the most important intent first
INTENT_DATA = {
    # Intent 3: The main action, prioritized first
    "Summarize": {
        "keywords": ["summarize", "text", "extract", "shorten", "gist"],
        "response": "Please paste the full text you want me to summarize now (or type 'stop' to cancel):"
    },

    # Intent 1: A general greeting
    "Greeting": {
        "keywords": ["hello", "hi", "hey", "greetings"],
        "response": "Hello there! I'm a simple text summarizer. How can I assist you today?"
    },

    # Intent 2: Asking for the chatbot's function or purpose
    "Function_Query": {
        "keywords": ["what", "do", "can", "your", "purpose", "function"],
        "response": "I am a simple Extractive Summarization tool. Give me a text, and I will try to select the most important sentence."
    }
}
# The default response if no intent is matched
DEFAULT_RESPONSE = "I'm sorry, I don't quite understand. Could you rephrase that?"

print("Intents and example queries (keywords) defined.")
```

```python
# CELL 2: Preprocessing Function

import re # Import the regular expression module

def preprocess_text(text):
    """
    Cleans the input text for easier keyword matching.
    """
    # 1. Convert to lowercase
    text = text.lower()

    # 2. Remove punctuation (optional, but good for simple keyword matching)
    # This keeps letters, numbers, and spaces
    text = re.sub(r'[^\w\s]', '', text)

    # 3. Remove extra whitespace
    text = re.sub(r'\s+', ' ', text).strip()

    # NOTE for a beginner project, we skip Stop Word Removal for simplicity

    return text

# Example Test
test_input = "  Hello! What can YOU do?"
processed_output = preprocess_text(test_input)
print(f"Original: '{test_input}'")
print(f"Processed: '{processed_output}'")
```

```python
# CELL 3: Core Logic - Intent Matching, Summarization, and Response Generation

def match_intent(processed_input):
    """
    Matches the processed input text to a defined intent using keywords.
    Returns the matched intent name or None.
    """
    # IMPORTANT: The order of iteration matters!
    # Put specific action intents (like Summarize) first.

    # We will use the order they are defined in INTENT_DATA:
    # 1. Greeting
    # 2. Function_Query
    # 3. Summarize

    for intent_name, data in INTENT_DATA.items():
        # Check if ANY of the intent's keywords are present in the processed input
        if any(keyword in processed_input for keyword in data["keywords"]):
            return intent_name
    return None

def generate_simple_summary(full_text):
    """
    A very basic Extractive Summarization function: it extracts the first sentence.
    """
    # Find the first period, question mark, or exclamation mark that ends a sentence
    sentence_endings = re.search(r'[.?!]', full_text)

    if sentence_endings:
        # Extract everything up to (and including) the sentence ending
        summary = full_text[:sentence_endings.end()].strip()
    else:
        # If no punctuation is found, just return the whole (short) text
        summary = full_text

    return "💡 **Summary (First Sentence Extracted):**\n" + summary


def get_response(user_input):
    """
    Main function to process input and generate the final response.
    """
    processed_input = preprocess_text(user_input)
    matched_intent = match_intent(processed_input)

    # Case 1: Intent Matched
    if matched_intent:
        response_template = INTENT_DATA[matched_intent]["response"]
        return matched_intent, response_template

    # Case 2: No Intent Matched
    else:
        return "Unknown", DEFAULT_RESPONSE

print("Intent matching and response generation functions defined.")
```

```python
# CELL 4: Test Scenarios and Main Chat Loop

print("--- Chatbot Simulation Started ---")

# --- Scenario 1: Greeting ---
test1 = "Hi, can you help me?"
intent1, response1 = get_response(test1)
print(f"\nUser: {test1}")
print(f"Intent: {intent1}")
print(f"Chatbot: {response1}")

# --- Scenario 2: Function Query ---
test2 = "What is your main function?"
intent2, response2 = get_response(test2)
print(f"\nUser: {test2}")
print(f"Intent: {intent2}")
print(f"Chatbot: {response2}")

# --- Scenario 3: Summarize Intent & Action ---
test3 = "I need you to summarize this text."
intent3, response3 = get_response(test3)
print(f"\nUser: {test3}")
print(f"Intent: {intent3}")
print(f"Chatbot: {response3}")

# Simulating the user providing the text after the 'Summarize' prompt
text_to_summarize = (
    "The sun is the star at the center of the Solar System. "
    "It is a nearly perfect sphere of hot plasma, heated to incandescence "
    "by nuclear fusion reactions in its core. The Sun radiates this energy mainly as light and heat."
)

summary_result = generate_simple_summary(text_to_summarize)
print(f"\nUser (Text Provided):\n{text_to_summarize}")
print(f"Chatbot:\n{summary_result}")

# --- Scenario 4: Unknown Input ---
test4 = "I want a cup of coffee."
intent4, response4 = get_response(test4)
print(f"\nUser: {test4}")
print(f"Intent: {intent4}")
print(f"Chatbot: {response4}")

print("\n--- Chatbot Simulation Ended ---")
```

```python
# CELL 5: Main Interactive Chat Loop

def start_chatbot():
    """
    Runs the main conversation loop, allowing the user to chat with the bot.
    """

    # Define exit command
    EXIT_COMMANDS = ["bye", "exit", "stop", "quit"]

    print("-------------------------------------------------------")
    print("💬 Chatbot Started! Type 'bye' or 'exit' to end the chat.")
    print("-------------------------------------------------------")

    # Start the continuous loop
    while True:
        # Get user input
        user_input = input("You: ")

        # Check for exit command
        if user_input.lower() in EXIT_COMMANDS:
            print("\nChatbot: Goodbye! Have a great day!")
            break

        # 1. Get the detected intent and the initial response template
        intent, response_template = get_response(user_input)

        # 2. Handle the specific 'Summarize' action
        if intent == "Summarize":
            # Respond with the prompt to paste the text
            print(f"Chatbot: {response_template}")

            # Get the text the user wants to summarize (next input)
            text_to_summarize = input("Paste Text Here: ")

            # Check if the user is trying to cancel the paste
            if text_to_summarize.lower() in EXIT_COMMANDS:
                print("Chatbot: Summarization cancelled. How else can I help?")
                continue # Go back to the main loop

            # Perform the extractive summarization
            summary = generate_simple_summary(text_to_summarize)

            # Print the summary result
            print(f"Chatbot:\n{summary}")

        # 3. Handle all other intents (Greeting, Function_Query, Unknown)
        else:
            print(f"Chatbot: {response_template}")

# Run the chatbot!
start_chatbot()
```
