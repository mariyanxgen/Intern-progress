```python
# Step 1: Install & Import
!pip install nltk

import nltk
from nltk import word_tokenize, pos_tag, ne_chunk
from nltk.tree import Tree

# Step 2: Download required NLTK data
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('maxent_ne_chunker')
nltk.download('maxent_ne_chunker_tab')
nltk.download('words')

# Step 3: Input Text (can be multi-line or a list of sentences)
texts = [
    "Apple is looking at buying U.K. startup for $1 billion.",
    "Elon Musk founded SpaceX in 2002 in California.",
    "Barack Obama was the 44th President of the United States.",
    "Google announced a new AI project in New York."
]

# Step 4: List of known companies for correction
companies = ["Apple", "Google", "Microsoft", "Amazon", "Facebook", "Meta"]

# Step 5: Function to extract NER from a sentence
def extract_entities(text):
    tokens = word_tokenize(text)
    pos_tags = pos_tag(tokens)
    ner_tree = ne_chunk(pos_tags)

    entities = []
    for chunk in ner_tree:
        if hasattr(chunk, 'label'):
            entity_name = " ".join(c[0] for c in chunk)
            entity_type = chunk.label()

            # Automatic correction for known companies
            if entity_name in companies:
                entity_type = "ORGANIZATION"

            entities.append((entity_name, entity_type))
    return entities, ner_tree

# Step 6: Process all sentences
all_entities = []
for sentence in texts:
    entities, ner_tree = extract_entities(sentence)
    all_entities.extend(entities)

    # Display text tree (Colab-friendly)
    print(f"\nSentence: {sentence}")
    print("NER Tree:")
    print(ner_tree.pformat())  # text tree visualization

# Step 7: Final extracted entities
print("\n\nAll Named Entities Found:")
for e in all_entities:
    print(e)
```
