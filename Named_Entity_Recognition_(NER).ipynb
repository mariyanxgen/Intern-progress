# 🧠 Named Entity Recognition (NER)

This notebook performs Named Entity Recognition (NER) using spaCy.  
NER is a natural language processing task that identifies entities such as names of people, organizations, locations, dates, etc., in text.

```python
import spacy
nlp = spacy.load("en_core_web_sm")
```

```python
import spacy
nlp = spacy.load("en_core_web_sm")
```

```python
text = """
Elon Musk founded SpaceX in 2002 and Tesla Motors in 2003.
He was born in South Africa and studied at the University of Pennsylvania.
Apple Inc. was founded by Steve Jobs, Steve Wozniak, and Ronald Wayne in 1976.
"""
```

```python
doc = nlp(text)

print("Named Entities, Phrases, and Labels:")
for entity in doc.ents:
    print(f"{entity.text} --> {entity.label_}")
```

```python
from spacy import displacy

displacy.render(doc, style="ent", jupyter=True)
```

```python
user_text = input("Enter a sentence for NER: ")
doc2 = nlp(user_text)

print("\nEntities found:")
for ent in doc2.ents:
    print(f"{ent.text} ({ent.label_})")

displacy.render(doc2, style="ent", jupyter=True)
```

```python
import pandas as pd

# Create a list of entities and labels
entities = [(ent.text, ent.label_) for ent in doc2.ents]

# Convert to DataFrame for clean display
df = pd.DataFrame(entities, columns=["Entity", "Label"])

print("Named Entities Detected:\n")
display(df)
```
