```python
# Step 1: Define intents and example queries

intents = {
    "greeting": {
        "keywords": ["hello", "hi", "hey", "good morning", "good evening"],
        "responses": ["Hello! How can I help you today?", "Hi there! What can I do for you?", "Hey! How’s it going?"]
    },
    "goodbye": {
        "keywords": ["bye", "goodbye", "see you", "take care"],
        "responses": ["Goodbye! Have a great day!", "See you soon!", "Take care!"]
    },
    "thanks": {
        "keywords": ["thank", "thanks", "thank you"],
        "responses": ["You're welcome!", "Glad to help!", "No problem!"]
    },
    "time": {
        "keywords": ["time", "clock", "current time"],
        "responses": ["Let me check the time for you..."]
    },
    "weather": {
        "keywords": ["weather", "temperature", "rain", "forecast"],
        "responses": ["I can’t fetch live weather data right now, but it looks like a good day!"]
    }
}
```

```python
# Step 2: Preprocess user input
import re

def preprocess_text(text):
    text = text.lower()  # lowercase
    text = re.sub(r'[^\w\s]', '', text)  # remove punctuation
    tokens = text.split()  # simple tokenization
    return tokens
```

```python
sample = "Hello, How are you?"
print(preprocess_text(sample))
```

```python
# Step 3: Match input to intents using keyword matching

import random
from datetime import datetime

def get_intent(user_input):
    tokens = preprocess_text(user_input)
    for intent, data in intents.items():
        for word in tokens:
            if word in data["keywords"]:
                return intent
    return "unknown"

def generate_response(intent):
    if intent == "time":
        return f"The current time is {datetime.now().strftime('%H:%M:%S')}."
    elif intent in intents:
        return random.choice(intents[intent]["responses"])
    else:
        return "I'm sorry, I didn’t quite understand that. Can you rephrase?"
```

```python
# Step 4: Chatbot conversation loop

def chatbot():
    print("Chatbot: Hello! I’m your simple assistant. Type 'exit' to end the chat.")
    while True:
        user_input = input("You: ")
        if user_input.lower() == "exit":
            print("Chatbot: Goodbye! ")
            break

        intent = get_intent(user_input)
        response = generate_response(intent)
        print("Chatbot:", response)

# Uncomment below to test in Colab
# chatbot()
```
