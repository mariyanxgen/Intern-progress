```python
# ======================================
# 🧠 Text Summarization (Extractive) Chatbot (Stable Version using sumy)
# ======================================

# ✅ Step 1: Install Libraries
!pip install nltk sumy

import re
import nltk
from sumy.parsers.plaintext import PlaintextParser
from sumy.nlp.tokenizers import Tokenizer
from sumy.summarizers.lex_rank import LexRankSummarizer

nltk.download('punkt')

# ✅ Step 2: Define Intents and Example Queries
intents = {
    "greet": {
        "patterns": ["hi", "hello", "hey", "good morning", "good evening"],
        "response": "Hello! 👋 I'm your Text Summarization Assistant. Please enter text to summarize."
    },
    "summarize": {
        "patterns": ["summarize", "shorten", "make it brief", "generate summary"],
        "response": "Sure! Please paste the paragraph or article you'd like me to summarize."
    },
    "bye": {
        "patterns": ["bye", "goodbye", "see you"],
        "response": "Goodbye! 👋 Have a great day!"
    }
}

# ✅ Step 3: Preprocess Input Text
def preprocess_text(text):
    text = text.lower()
    text = re.sub(r'[^a-z0-9\s\.\,]', '', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text

# ✅ Step 4: Match Input to Intents using Regex / Keywords
def identify_intent(user_input):
    user_input = user_input.lower()
    for intent, intent_data in intents.items():
        for pattern in intent_data['patterns']:
            if re.search(pattern, user_input):
                return intent
    return "summarize"  # default if not matched

# ✅ Step 5: Generate Responses (with LexRank Summarizer)
def summarize_text(text, sentence_count=3):
    parser = PlaintextParser.from_string(text, Tokenizer("english"))
    summarizer = LexRankSummarizer()
    summary_sentences = summarizer(parser.document, sentence_count)
    return " ".join(str(sentence) for sentence in summary_sentences)

def chatbot_response(user_input):
    intent = identify_intent(user_input)

    if intent in ["greet", "bye"]:
        return intents[intent]["response"]

    elif intent == "summarize":
        if len(user_input.split()) < 30:
            return "Please provide a longer text (at least 30 words) for summarization."
        try:
            summary = summarize_text(user_input)
            return f"📄 Summary:\n\n{summary if summary else 'Unable to generate summary — text too short.'}"
        except Exception as e:
            return f"⚠️ Error during summarization: {str(e)}"

    else:
        return "Sorry, I didn’t understand that."

# ✅ Step 6: Test Chatbot on Multiple Scenarios
test_inputs = [
    "Hi there!",
    "Can you summarize this text?",
    """Artificial Intelligence (AI) is a rapidly advancing field that aims to create machines
    capable of performing tasks that typically require human intelligence. These tasks include learning, reasoning,
    problem-solving, perception, and language understanding. AI applications are seen in industries such as healthcare,
    finance, transportation, and entertainment. In healthcare, AI helps detect diseases early, while in finance it assists
    in fraud detection. As AI continues to grow, it raises both opportunities and ethical challenges.""",
    "Bye!"
]

print("🧩 Testing Chatbot Responses:\n")
for inp in test_inputs:
    print(f"👤 User: {inp}")
    print(f"🤖 Bot: {chatbot_response(inp)}\n")

# ✅ Step 7: Save Script / Notebook
with open("Text_Summarization_Chatbot_sumy.py", "w") as f:
    f.write("# Extractive Text Summarization Chatbot using LexRank (sumy)\n")
print("✅ Notebook completed and script saved as 'Text_Summarization_Chatbot_sumy.py'.")
```
